# Datenvisualisierung Dashboard

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd


%matplotlib inline


def mplplot(df, **kwargs):
    fig = df.plot().get_figure()
    plt.close(fig)
    return fig


def sine(frequency=1.0, amplitude=1.0, n=200, view_fn=mplplot):
    xs = np.arange(n) / n * 20.0
    ys = amplitude * np.sin(frequency * xs)
    df = pd.DataFrame(dict(y=ys), index=xs)
    return view_fn(df, frequency=frequency, amplitude=amplitude, n=n)
```

```python
import hvplot.pandas


def hvplot(df, **kwargs):
    return df.hvplot()


pn.interact(sine, view_fn=hvplot)
```

Sinuskurve, die interaktiv verändert werden kann, genauer kann

1. mit frequency die Frequenz geändert werden
2. mit amplitude ändert sich die Höhe des Ausschlags oder die Skalierung der y-Achse
3. mit n wird die Anzahl der Punkte definiert, aus der die Sinuskurve berechnet werden soll
